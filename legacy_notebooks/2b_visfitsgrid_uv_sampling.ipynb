```python
## sky signal ####
loc='/home/sukhdeep/Documents/Research/Bispectrum/Visibility_Bispectrum/Visibility_Correlation/asif_da/2b_visfitsgrid/sampling_2048'

NN = 2048  # Example value, replace with the actual NN value
num_elements = NN * (NN // 2 + 1) * 2  # Two doubles (real and imaginary) per element
data = np.fromfile(loc, dtype=np.float64, count=num_elements)
real_part=data.reshape(NN, (NN // 2 + 1))
```

```python
plt.imshow(real_part)
```

```python
full_array = np.zeros((NN, NN), dtype=np.float64)
full_array[:, :(NN // 2 + 1)] = real_part

for i in range(NN):
    for j in range(1, NN // 2):
        full_array[-i, -j] = full_array[i, j]
```

```python
xx=np.fft.fftshift(full_array)
```

```python
plt.imshow(xx,origin='lower',extent=[-NN//2,NN//2,-NN//2,NN//2],cmap='viridis', norm=LogNorm())  #, vmin=0,vmax=1
# plt.imshow(xx,origin='lower',extent=[-NN//2,NN//2,-NN//2,NN//2],cmap='viridis', vmin=0,vmax=1)  #, vmin=0,vmax=1

plt.colorbar()
plt.xlim(-200,200)
plt.ylim(-200,200)
```

```python
xx[xx==0]=-5000
plt.imshow(xx,origin='lower',extent=[-NN//2,NN//2,-NN//2,NN//2],cmap='tab20c')  #, vmin=0,vmax=1
plt.colorbar()
plt.xlim(-100,100)
plt.ylim(-100,100)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap, BoundaryNorm
from matplotlib.colors import LogNorm

# Define the colormap
cmap = ListedColormap(['white', 'black',  'yellow','green','blue', 'violet', 'red'])

# Define the boundaries
bounds = [0, 1, 10, 30, 50, 100, np.max(xx)]

# Create the normalization
norm = BoundaryNorm(bounds, cmap.N)

# Plot with imshow
# plt.imshow(full_array, origin='lower', extent=[-NN//2, NN//2, -NN//2, NN//2], cmap=cmap, norm=norm)
# plt.imshow(xx, origin='lower', extent=[-NN//2, NN//2, -NN//2, NN//2])
plt.imshow(xx, origin='lower', extent=[-NN//2, NN//2, -NN//2, NN//2], cmap=cmap, norm=norm)
plt.colorbar()
plt.title('UV sampling')

plt.xlim(-300,300)
plt.ylim(-300,300)

# plt.savefig('uv_sampling.png', format='png', dpi=300)
```
