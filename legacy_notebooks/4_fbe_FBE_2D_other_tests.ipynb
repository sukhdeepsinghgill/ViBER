```python
# du_ring=(Nmax-1)/(1.*Nr) 
# uLow=np.array([i*du_ring+1 for i in range(Nr)]  )
# uHigh=np.array([(i+1)*du_ring+1 for i in range(Nr)])
# uavg=(2/3)*(uHigh**3-uLow**3)/(uHigh**2-uLow**2)
```

```python
Nbin = 31
Nmax = 660
Nr=Nbin

NN=2048
LL=3.438/2
LL*=np.pi/(60*180)
Cxy = 1/(NN*LL)

currentNumber = 1
uLow = [0] * Nbin
uHigh = [0] * Nbin

# Define intervals
# intervals = [(0, 50, 5), (50, 200, 10), (200, 350, 50)]
intervals = [(0, 50, 5), (50, 100, 10),(100, 200, 20), (200, 410, 35), (410, 660, 50)]

# Index for uLow array
index = 0

for start, end, step in intervals:
    for i in range(start, end, step):
        if index < Nbin:
            uLow[index] = i
            if index > 0:
                uHigh[index - 1] = uLow[index]
            index += 1

# Handle the last uHigh value if it hasn't been set
if index > 0 and index < Nbin:
    uHigh[index - 1] = Nmax

uLow[0] = 1   
uHigh[Nbin - 1] = 1.0 * Nmax

uavg=(2/3)*(np.array(uHigh)**3-np.array(uLow)**3)/(np.array(uHigh)**2-np.array(uLow)**2)
# Print the arrays for verification
print("uLow:", uLow)
print("uHigh:", uHigh)
print("Diff:", np.array(uHigh)-np.array(uLow))
```
